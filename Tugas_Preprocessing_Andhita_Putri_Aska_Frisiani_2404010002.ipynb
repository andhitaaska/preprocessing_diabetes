```python
df = pd.read_csv("/content/diabetes.csv")

print("Data berhasil dibaca")
print("Jumlah baris:", len(df))
print("Jumlah kolom:", len(df.columns))

df.head()
```

```python
# Menyimpan salinan data asli
df_awal = df.copy()

# Mengubah tanda ? menjadi data kosong
df = df.replace("?", np.nan)

print("Jumlah data kosong:")
print(df.isnull().sum())

print("\nJumlah seluruh data kosong:")
print(df.isnull().sum().sum())

print("\nJumlah baris duplikat:")
print(df.duplicated().sum())
```

```python
# Memisahkan kolom numerik dan kategorikal
kolom_numerik = df.select_dtypes(
    include=["number"]
).columns

kolom_kategori = df.select_dtypes(
    include=["object", "category", "bool"]
).columns

# Mengisi data kosong pada kolom numerik
for kolom in kolom_numerik:
    if df[kolom].isnull().any():
        median = df[kolom].median()

        if pd.notna(median):
            df[kolom] = df[kolom].fillna(median)

# Mengisi data kosong pada kolom kategorikal
for kolom in kolom_kategori:
    if df[kolom].isnull().any():
        modus = df[kolom].mode()

        if not modus.empty:
            df[kolom] = df[kolom].fillna(modus[0])

print("Proses pengisian data selesai")
```

```python
print("Data kosong sebelum preprocessing:")
print(df_awal.replace("?", np.nan).isnull().sum().sum())

print("\nData kosong setelah preprocessing:")
print(df.isnull().sum().sum())

print("\nJumlah baris duplikat:")
print(df.duplicated().sum())

display(df.head())
```

```python
print("DATA SEBELUM PREPROCESSING")
print("Jumlah baris:", df_awal.shape[0])
print("Jumlah kolom:", df_awal.shape[1])

print("\nDATA SETELAH PREPROCESSING")
print("Jumlah baris:", df.shape[0])
print("Jumlah kolom:", df.shape[1])
```

```python
df.to_csv(
    "/content/diabetes_hasil_preprocessing.csv",
    index=False
)

print("Dataset hasil preprocessing berhasil disimpan")
```
